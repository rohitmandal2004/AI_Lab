```python
from queue import PriorityQueue

# Graph Representation (Adjacency List)
graph = {
    'S': [('A', 1), ('E', 4)],
    'A': [('B', 2), ('C', 5)],
    'B': [('D', 3), ('G', 4)],
    'C': [('D', 2)],
    'D': [('G', 2)],
    'E': [('G', 6)],
    'G': []
}

# Heuristic values
heuristic = {
    'S': 7,
    'A': 6,
    'B': 4,
    'C': 4,
    'D': 2,
    'E': 6,
    'G': 0
}

# Function to print graph
print("Graph Representation:\n")
for node, neighbors in graph.items():
    print(f"{node} -> {neighbors}")

print("\nHeuristic Values:")
for node in heuristic:
    print(f"h({node}) = {heuristic[node]}")


def h(node):
    return heuristic[node]


def get_neighbors(node):
    return graph[node]


def a_star(start, goal):
    open_set = PriorityQueue()
    open_set.put((h(start), start))

    came_from = {}

    g_score = {start: 0}
    f_score = {start: h(start)}

    while not open_set.empty():

        current = open_set.get()[1]

        print("\nCurrent Node:", current)

        if current == goal:
            path = [current]
            while current in came_from:
                current = came_from[current]
                path.append(current)
            path.reverse()

            print("\nShortest Path:", path)
            print("Total Cost:", g_score[goal])
            return path

        for neighbor, cost in get_neighbors(current):

            print("Neighbor:", neighbor)
            print("Edge Cost:", cost)

            tentative_g = g_score[current] + cost
            print("Tentative g =", tentative_g)

            print("Heuristic =", h(neighbor))

            tentative_f = tentative_g + h(neighbor)
            print("f =", tentative_f)

            if neighbor not in g_score or tentative_g < g_score[neighbor]:
                came_from[neighbor] = current
                g_score[neighbor] = tentative_g
                f_score[neighbor] = tentative_f
                open_set.put((tentative_f, neighbor))

    print("No Path Found!")
    return None


# Driver Code
start = 'S'
goal = 'G'

print("\nStart Node:", start)
print("Goal Node:", goal)

a_star(start, goal)
```
