```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Name": ["Rohit" , "Rishi", "Sobuj"],
    "Marks": [87,99,78],
    "Department": ["CSE", "AIML", "DS"]

}

df = pd.DataFrame(data)
```

```python
df
```

```python
#create a bar chart showing student marks 
plt.bar(df["Name"], df["Marks"])
plt.title("Student Marks chart")
plt.xlabel("Student Name")
plt.ylabel("Marks")
```

```python
plt.pie(
    df["Marks"],
    labels = df["Name"],
    autopct = "%1.1f%%"
)
```
