```python
# Example filter: rows where PM2.5 is above 100
filtered_df = df[df["PM2.5"] > 100]

# Example filter: rows for a specific country
filtered_by_country = df[df["Country"] == "India"]

filtered_df.head()
```

```python
# Various example filters on existing `df`
# (uses pd and df already defined in earlier cells)

date_series = pd.to_datetime(df["Date"])

# 1) Date range filter
by_date_range = df[(date_series >= "2023-01-01") & (date_series <= "2023-06-30")]

# 2) High PM2.5
high_pm25 = df[df["PM2.5"] > 100]

# 3) Moderate PM2.5 (50-100)
moderate_pm25 = df[(df["PM2.5"] > 50) & (df["PM2.5"] <= 100)]

# 4) Specific cities
cities = ["Mumbai", "Bangkok", "Beijing"]
city_filter = df[df["City"].isin(cities)]

# 5) Top 10 rows by PM2.5
top10_pm25 = df.nlargest(10, "PM2.5")

# 6) Low temperature but high humidity
cold_humid = df[(df["Temperature"] < 0) & (df["Humidity"] > 80)]

# 7) Combined condition: India and high PM10
india_high_pm10 = df[(df["Country"] == "India") & (df["PM10"] > 100)]

# 8) Exclude a country
not_china = df[df["Country"] != "China"]

# Quick summary outputs
for name, d in [
    ("by_date_range", by_date_range),
    ("high_pm25", high_pm25),
    ("moderate_pm25", moderate_pm25),
    ("city_filter", city_filter),
    ("top10_pm25", top10_pm25),
    ("cold_humid", cold_humid),
    ("india_high_pm10", india_high_pm10),
    ("not_china", not_china),
]:
    print(f"{name}: shape={d.shape}")
    display(d.head())
```

```python
df.isnull()
```

```python
df['City'] = df['City'].astype(str).str.strip()
df['Country'] = df['Country'].astype(str).str.strip()
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

numeric_cols = ['PM2.5', 'PM10', 'NO2', 'SO2', 'CO', 'O3', 'Temperature', 'Humidity', 'Wind Speed']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df = df.drop_duplicates().dropna(subset=['Date']).reset_index(drop=True)

print("shape:", df.shape)
print(df.isna().sum())
df.head()
```
